## Look at the answers

```
import json, glob
for path in sorted(glob.glob('eval/generation/*holdout.jsonl')):
    print('\n==', path)
    for line in open(path):
        r = json.loads(line)
        flag = 'OK ' if r['correct'] else 'MISS'
        print(flag, r['id'], r['question'], '->', r['answer'][:160])
```
